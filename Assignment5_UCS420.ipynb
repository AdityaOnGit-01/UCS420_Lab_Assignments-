```python

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)
print("Addition of 2:", arr + 2)
print("Multiplication by 3:", arr * 3)
print("Division by 2:", arr / 2)
```

```python
#a)
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
#b)
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

# For array x
values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("For x:")
print("Most frequent value:", most_frequent)
print("Indices:", np.where(x == most_frequent)[0])

# For array y
values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]

print("\nFor y:")
print("Most frequent value:", most_frequent)
print("Indices:", np.where(y == most_frequent)[0])
```

```python

import numpy as np

Aditya_Sharma = np.linspace(10, 100, 25)

print("Array:", Aditya_Sharma)
print("Dimensions:", Aditya_Sharma.ndim)
print("Shape:", Aditya_Sharma.shape)
print("Total elements:", Aditya_Sharma.size)
print("Data type:", Aditya_Sharma.dtype)
print("Total bytes:", Aditya_Sharma.nbytes)

# Transpose using reshape
transposed = Aditya_Sharma.reshape(1, 25)
print("Transpose using reshape:")
print(transposed)

# Transpose using T
print("Transpose using T:")
print(Aditya_Sharma.T)
```

```python

import numpy as np

ucs420_Aditya_Sharma = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array:")
print(ucs420_Aditya_Sharma)

print("Mean:", np.mean(ucs420_Aditya_Sharma))
print("Median:", np.median(ucs420_Aditya_Sharma))
print("Maximum:", np.max(ucs420_Aditya_Sharma))
print("Minimum:", np.min(ucs420_Aditya_Sharma))
print("Unique elements:", np.unique(ucs420_Aditya_Sharma))

# Reshape into 4 rows and 3 columns
reshaped_ucs420_Aditya_Sharma = ucs420_Aditya_Sharma.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_Aditya_Sharma)

# Resize into 2 rows and 3 columns
resized_ucs420_Aditya_Sharma = np.resize(ucs420_Aditya_Sharma, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_Aditya_Sharma)
```
