```python
!pip install -q easyocr opencv-python-headless

import cv2
import re
import easyocr
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
image_path = next(iter(uploaded))

image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(7, 5))
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Medicine Bottle Image")
plt.show()

reader = easyocr.Reader(['en'])
results = reader.readtext(image_path)

text = " ".join([item[1] for item in results])

print("Extracted Text:")
print(text)

medicine = "Not detected"
dosage = "Not detected"
instructions = "Not detected"

lines = [item[1] for item in results]

if lines:
    medicine = lines[0]

match = re.search(r'\b\d+(?:\.\d+)?\s*(mg|ml|mcg|g)\b', text, re.I)

if match:
    dosage = match.group()

for line in lines:
    if re.search(r'after food|before food|twice daily|once daily|three times daily', line, re.I):
        instructions = line
        break

print("\nMEDICINE READING ASSISTANT")
print("Medicine Name:", medicine)
print("Dosage:", dosage)
print("Instructions:", instructions)
print("\nPlease verify all information with the original medicine label.")
```

```python
!pip install -q deepface tf-keras opencv-python-headless

import cv2
import matplotlib.pyplot as plt
from deepface import DeepFace
from google.colab import files

uploaded = files.upload()
image_path = next(iter(uploaded))

image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

result = DeepFace.analyze(
    img_path=image_path,
    actions=['emotion'],
    enforce_detection=True
)

emotion = result[0]['dominant_emotion']
confidence = result[0]['emotion'][emotion]

responses = {
    'happy': "You seem to have a positive expression. Keep smiling!",
    'sad': "You seem to have a sad expression. Take a break and relax.",
    'angry': "You seem to have an angry expression. Take a deep breath.",
    'fear': "You seem worried. Take things one step at a time.",
    'surprise': "You seem surprised. Hope it is a good surprise!",
    'disgust': "You seem uncomfortable. Take a moment to relax.",
    'neutral': "Your expression appears neutral. Have a productive day!"
}

plt.figure(figsize=(7, 5))
plt.imshow(image_rgb)
plt.axis('off')
plt.title("Facial Expression Detection")
plt.show()

print("EMOTION-AWARE COGNITIVE ASSISTANT")
print("--------------------------------")
print("Detected Expression:", emotion.capitalize())
print("Confidence:", round(confidence, 2), "%")
print("\nAssistant Response:")
print(responses[emotion])
```
