```python

L = [10, 20, 30, 40, 50, 60, 70, 80]

# i. Add 200 and 300
L.append(200)
L.append(300)
print("After adding:", L)

# ii. Remove 10 and 30
L.remove(10)
L.remove(30)
print("After removing:", L)

# iii. Sort in ascending order
L.sort()
print("Ascending order:", L)

# iv. Sort in descending order
L.sort(reverse=True)
print("Descending order:", L)
```

```python

scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

# i. Highest score and its index
highest = max(scores)
print("Highest score:", highest)
print("Index of highest score:", scores.index(highest))

# ii. Lowest score and its count
lowest = min(scores)
print("Lowest score:", lowest)
print("Count of lowest score:", scores.count(lowest))

# iii. Reverse the tuple and convert it to a list
reversed_scores = list(scores[::-1])
print("Reversed list:", reversed_scores)

# iv. Search for a score entered by the user
x = float(input("Enter a score to search: "))

if x in scores:
    print("Score found at index:", scores.index(x))
else:
    print("Score is not present")
```

```python

import random

numbers = []

# Generate 100 random numbers
for i in range(100):
    n = random.randint(100, 900)
    numbers.append(n)

odd = []
even = []
prime = []

# Check odd and even numbers
for n in numbers:
    if n % 2 == 0:
        even.append(n)
    else:
        odd.append(n)

    # Check whether the number is prime
    is_prime = True

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

    if is_prime:
        prime.append(n)

print("Random numbers:", numbers)
print("Odd numbers:", odd)
print("Total odd numbers:", len(odd))

print("Even numbers:", even)
print("Total even numbers:", len(even))

print("Prime numbers:", prime)
print("Total prime numbers:", len(prime))
```

```python

A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

# i. Union of both sets
print("Union:", A.union(B))

# ii. Common scores
print("Intersection:", A.intersection(B))

# iii. Scores exclusive to each team
print("Symmetric difference:", A.symmetric_difference(B))

# iv. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# v. Remove a score entered by the user from set A
x = int(input("Enter a score to remove from A: "))

if x in A:
    A.remove(x)
    print("Updated set A:", A)
else:
    print("Score is not present in set A")
```
