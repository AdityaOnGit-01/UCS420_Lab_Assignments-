```python

import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# My college roll number
roll_no = "1024170332"

# Take the last two digits
digits = roll_no[-2:]

categories = ["billing", "account", "general"]

extra_entries = []

for digit in digits:
    d = int(digit)
    category = categories[d % 3]

    if category == "billing":
        entry = {
            "question": "how can i check my payment status",
            "answer": "You can check payment status in the billing section.",
            "keywords": "payment status bill",
            "category": category
        }
    elif category == "account":
        entry = {
            "question": "how can i update my email address",
            "answer": "Open account settings and update your email address.",
            "keywords": "email update account",
            "category": category
        }
    else:
        entry = {
            "question": "where can i get help",
            "answer": "You can contact our support team for help.",
            "keywords": "help support information",
            "category": category
        }

    extra_entries.append(entry)

# Combine all six entries
all_entries = fixed_entries + extra_entries

df = pd.DataFrame(all_entries)

print(df)
```

```python

def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for index, row in df.iterrows():
        text = (row["question"] + " " + row["answer"] + " " +
                row["keywords"]).lower()

        matches = 0
        for word in query_words:
            if word in text:
                matches += 1

        if matches > 0:
            confidence = matches / len(query_words) * 100
            results.append((index, confidence))

    results.sort(key=lambda x: x[1], reverse=True)

    for index, confidence in results:
        print("Question:", df.loc[index, "question"])
        print("Category:", df.loc[index, "category"])
        print("Confidence:", round(confidence, 2), "%")
        print()

    if len(results) == 0:
        print("No matching FAQ found.")


score_query("fee payment", df)
```

```python

def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for index, row in df.iterrows():
        text = (row["question"] + " " + row["answer"] + " " +
                row["keywords"]).lower()

        matches = 0
        for word in query_words:
            if word in text:
                matches += 1

        if matches > 0:
            confidence = matches / len(query_words) * 100
            results.append((index, confidence))

    if len(results) == 0:
        print("No matching FAQ found.")
        return

    results.sort(key=lambda x: x[1], reverse=True)

    highest_score = results[0][1]
    best_matches = []

    for index, score in results:
        if score == highest_score:
            best_matches.append((index, score))

    if len(best_matches) > 1:
        print("Tie found! All highest-scoring matches:")
    else:
        print("Best match:")

    for index, score in best_matches:
        print("Question:", df.loc[index, "question"])
        print("Category:", df.loc[index, "category"])
        print("Confidence:", round(score, 2), "%")
        print()


# Query that creates a tie between the two fixed fee questions
score_query("fee", df)

# Query with one highest-scoring entry
score_query("password", df)
```
