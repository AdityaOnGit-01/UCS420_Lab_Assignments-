```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from collections import Counter
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = """Artificial intelligence is my favorite topic because it is changing the world.
It helps people solve complex problems and makes everyday tasks easier.
AI is used in healthcare, education, business, and many other fields.
I enjoy learning about machine learning and how computers understand data.
I believe artificial intelligence will create many new opportunities in the future."""

clean_text = re.sub(r'[^\w\s]', '', text.lower())

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("Cleaned Text:")
print(clean_text)

print("\nWords using split():")
print(clean_text.split())

print("\nWords using word_tokenize():")
print(words)

print("\nSentences:")
print(sentences)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("\nWords after removing stopwords:")
print(filtered_words)

frequency = Counter(filtered_words)

print("\nWord Frequency:")
print(frequency)

plt.bar(frequency.keys(), frequency.values())
plt.xticks(rotation=45)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Word Frequency Distribution")
plt.show()
```

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

text = """Artificial intelligence is my favorite topic because it is changing the world.
It helps people solve complex problems and makes everyday tasks easier.
AI is used in healthcare, education, business, and many other fields.
I enjoy learning about machine learning and how computers understand data.
I believe artificial intelligence will create many new opportunities in the future."""

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Original Words:")
print(filtered_words)

print("\nStemmed Words:")
print(stemmed_words)

print("\nLemmatized Words:")
print(lemmatized_words)

print("\nComparison:")
for original, stem, lemma in zip(filtered_words, stemmed_words, lemmatized_words):
    print(original, "->", stem, "->", lemma)
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Artificial intelligence helps computers learn and solve problems.",
    "Electric vehicles reduce pollution and save fuel.",
    "Machine learning helps businesses analyze data and make decisions."
]

cv = CountVectorizer()
bow = cv.fit_transform(texts)

print("Bag of Words:")
print(cv.get_feature_names_out())
print(bow.toarray())

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(texts)

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

features = tfidf.get_feature_names_out()

for i, text in enumerate(texts):
    scores = tfidf_matrix[i].toarray()[0]
    top_indices = scores.argsort()[-3:][::-1]

    print("\nText", i + 1)
    print("Top 3 Keywords:")

    for index in top_indices:
        print(features[index], ":", round(scores[index], 3))
```

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """Artificial intelligence allows computers to learn from data.
It helps machines solve problems and make decisions.
AI is widely used in healthcare and education.
It improves automation and saves time.
Artificial intelligence is becoming important in modern technology."""

text2 = """Blockchain is a technology used to store digital information.
It maintains secure and decentralized records.
Blockchain is widely used in cryptocurrency and banking.
It improves transparency and data security.
This technology is becoming important in modern businesses."""

words1 = set(re.findall(r'\b[a-zA-Z]+\b', text1.lower()))
words2 = set(re.findall(r'\b[a-zA-Z]+\b', text2.lower()))

jaccard = len(words1.intersection(words2)) / len(words1.union(words2))

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(vectors[0:1], vectors[1:2])[0][0]

print("Jaccard Similarity:", round(jaccard, 4))
print("Cosine Similarity:", round(cosine, 4))

if cosine > jaccard:
    print("Cosine similarity gives better insights in this case.")
else:
    print("Jaccard similarity gives better insights in this case.")
```

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

nltk.download('vader_lexicon')

reviews = [
    "The product is amazing and works perfectly. I really love it.",
    "Excellent quality and fast delivery. Very happy with the service.",
    "The product is terrible and completely disappointing.",
    "Very poor quality. I am unhappy with this purchase.",
    "The product is okay and works as expected.",
    "The service was average and nothing special."
]

sia = SentimentIntensityAnalyzer()
positive_reviews = []

for review in reviews:
    score = sia.polarity_scores(review)['compound']

    if score >= 0.05:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif score <= -0.05:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("Review:", review)
    print("Polarity:", score)
    print("Sentiment:", sentiment)
    print()

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white'
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Positive Reviews Word Cloud")
plt.show()
```

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """Artificial intelligence is changing the modern world.
Technology helps people solve problems and improve their daily lives.
Machine learning allows computers to learn from data and make predictions.
Deep learning is used in healthcare, education, and business.
Computers can recognize images, understand language, and process information.
Artificial intelligence is making machines smarter and more useful.
Technology continues to develop and create new opportunities.
Learning about artificial intelligence is important for future generations.
Data plays an important role in machine learning.
Modern computers can perform many complex tasks."""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

sequences = []

for line in text.split('.'):
    sequence = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(sequence)):
        sequences.append(sequence[:i + 1])

max_len = max(len(seq) for seq in sequences)

sequences = pad_sequences(
    sequences,
    maxlen=max_len,
    padding='pre'
)

X = sequences[:, :-1]
y = sequences[:, -1]

model = Sequential([
    Embedding(total_words, 50, input_length=max_len - 1),
    LSTM(100),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.fit(X, y, epochs=150, verbose=1)

seed_text = "artificial intelligence"

for i in range(25):
    sequence = tokenizer.texts_to_sequences([seed_text])[0]
    sequence = pad_sequences([sequence], maxlen=max_len - 1, padding='pre')

    prediction = model.predict(sequence, verbose=0)
    predicted_word = tokenizer.index_word[np.argmax(prediction)]

    seed_text += " " + predicted_word

print("Generated Text:")
print(seed_text)
```
