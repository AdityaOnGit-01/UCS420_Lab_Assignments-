```python

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading
corrected = temperature + 2
print("Corrected temperatures:", corrected)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected + 32
print("Temperatures in Fahrenheit:", fahrenheit)

# c. Find readings greater than 32°C
readings = corrected[corrected > 32]
print("Readings greater than 32°C:", readings)

# d. Count readings greater than 32°C
count = np.sum(corrected > 32)
print("Number of readings greater than 32°C:", count)

# e. Explanation
print("Vectorization performs operations on the whole array at once.")
print("Boolean indexing selects values that satisfy a condition.")
print("Both make code shorter and can be faster than using loops.")
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps recorded
print("Total steps:", np.sum(steps))

# b. Mean number of steps
print("Mean steps:", np.mean(steps))

# c. Maximum and minimum steps
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# d. Total steps for each day (axis=0)
print("Total steps for each day:", np.sum(steps, axis=0))

# e. Total steps for each user (axis=1)
print("Total steps for each user:", np.sum(steps, axis=1))

# f. Position of maximum number of steps
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum step value position (user, day):", position)
print("Maximum step value:", steps[position])
```

```python
import numpy as np

# a. Create the original array
original = np.array([1, 2, 3, 4, 5, 6])
print("Original array:", original)

# b. Create a slice from index 1 to 4
subset = original[1:5]
print("Subset:", subset)

# c. Change the first element of subset
subset[0] = 999
print("Original after changing subset:", original)
print("Modified subset:", subset)

# d. Create a copy and change its first element
copied = original[1:5].copy()
copied[0] = 500
print("Original after changing copy:", original)
print("Copied array:", copied)

# e. Create a 3x4 matrix using arange and reshape
matrix = np.arange(1, 13).reshape(3, 4)
print("3x4 matrix:\n", matrix)

# f. Extract rows, columns and selected elements
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

# g. Flatten the matrix using flatten() and ravel()
flat_array = matrix.flatten()
ravel_array = matrix.ravel()
print("Flatten array:", flat_array)
print("Ravel array:", ravel_array)

# h. Modify an element in ravel array
ravel_array[0] = 100
print("Matrix after modifying ravel:", matrix)

# i. Modify an element in flatten array
flat_array[1] = 200
print("Matrix after modifying flatten:", matrix)

# j. Display matrix properties
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python

import numpy as np

# Given data
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Display shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b. Calculate transpose
XT = X.T
print("\nTranspose of X:\n", XT)

# c. Calculate X transpose multiplied by X
XTX = XT @ X
print("\nX.T @ X:\n", XTX)

# d. Calculate the inverse
inverse = np.linalg.inv(XTX)
print("\nInverse of X.T @ X:\n", inverse)

# e. Calculate OLS coefficients
beta = inverse @ XT @ y
print("\nOLS coefficients:", beta)

# f. Display the meaning of coefficients
print("\nSleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# g. Predict the score for a new user
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\nPredicted assistance score:", predicted_score)
```
